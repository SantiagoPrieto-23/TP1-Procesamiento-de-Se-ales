# Trabajo Práctico 1: Procesamiento Digital de Señales
## Laboratorio de Análisis de Sistemas LTI, Filtrado y Coherencia

Este Notebook unifica las herramientas de análisis en el tiempo y la frecuencia, el diseño de filtros FIR básicos y el posterior análisis cruzado estadístico mediante la estimación de la **Función de Coherencia Cuadrática**.

### 1. Importación de Librerías y Definición de Funciones Básicas

```python
import numpy as np
import matplotlib.pyplot as plt

def generar_tonos_ruido(frecuencias, fs, duracion, amp_ruido):
    t = np.arange(0, duracion, 1/fs)
    senal = np.zeros_like(t)
    for f in frecuencias:
        senal += np.sin(2 * np.pi * f * t)
    ruido = amp_ruido * np.random.randn(len(t))
    return t, senal + ruido

def filtrar_tiempo(x, h):
    return np.convolve(x, h, mode='same')

def filtrar_frecuencia(x, h):
    N = len(x)
    h_pad = np.zeros(N)
    h_pad[:len(h)] = h
    X_w = np.fft.fft(x)
    H_w = np.fft.fft(h_pad)
    return np.real(np.fft.ifft(X_w * H_w))
```

### 2. Estimación de Coherencia (Método de Welch)
Para calcular correctamente la coherencia sin que dé trivialmente $1$, implementamos la segmentación, ventaneo con ventana *Hanning* y el promediado estadístico de los periodogramas cruzados.

```python
def calcular_densidades_espectrales(x, y, N_fft=1024, noverlap=512):
    N = len(x)
    step = N_fft - noverlap
    win = np.hanning(N_fft)
    win_norm = np.sum(win**2)
    
    Gxx_acum = np.zeros(N_fft, dtype=complex)
    Gyy_acum = np.zeros(N_fft, dtype=complex)
    Gxy_acum = np.zeros(N_fft, dtype=complex)
    cant_bloques = 0
    
    for start in range(0, N - N_fft + 1, step):
        end = start + N_fft
        x_chunk = x[start:end] * win
        y_chunk = y[start:end] * win
        
        X_w = np.fft.fft(x_chunk)
        Y_w = np.fft.fft(y_chunk)
        
        Gxx_acum += (X_w * np.conj(X_w)) / win_norm
        Gyy_acum += (Y_w * np.conj(Y_w)) / win_norm
        Gxy_acum += (Y_w * np.conj(X_w)) / win_norm
        cant_bloques += 1
        
    Gxx = np.real(Gxx_acum / cant_bloques)
    Gyy = np.real(Gyy_acum / cant_bloques)
    Gxy = Gxy_acum / cant_bloques
    
    f_norm = np.fft.fftfreq(N_fft)
    mitad = N_fft // 2
    return f_norm[:mitad], Gxx[:mitad], Gyy[:mitad], Gxy[:mitad]

def analizar_coherencia_sistema(x, y, fs=1.0, N_fft=1024, noverlap=512):
    f_norm, Gxx, Gyy, Gxy = calcular_densidades_espectrales(x, y, N_fft, noverlap)
    f_eje = f_norm * fs
    epsilon = 1e-12
    coherencia = (np.abs(Gxy)**2) / (Gxx * Gyy + epsilon)
    
    plt.figure(figsize=(10, 4))
    plt.plot(f_eje, coherencia, color='purple', linewidth=2, label=r'$\gamma_{xy}^2(\omega)$')
    plt.axhline(1.0, color='red', linestyle='--', alpha=0.5, label='Linealidad Perfecta')
    plt.title("Función de Coherencia Cuadrática")
    plt.xlabel("Frecuencia [Hz]" if fs != 1.0 else "Frecuencia Normalizada")
    plt.ylabel("Coherencia")
    plt.ylim(-0.05, 1.05)
    plt.grid(True, linestyle='--')
    plt.legend()
    plt.show()
    return f_eje, coherencia
```

### 3. Celda de Prueba Interactiva
Podemos simular un caso de estudio: una señal que pasa por un filtro de media móvil a la que se le suma ruido de medición en la salida, observando cómo se degrada la coherencia en ciertas bandas.

```python
# Parámetros de simulación
fs = 1000.0
t, x = generar_tonos_ruido(frecuencias=[50, 120, 250], fs=fs, duracion=2.0, amp_ruido=0.2)

# Respuesta del filtro LTI
h = np.ones(10) / 10.0
y_ideal = filtrar_tiempo(x, h)

# Añadimos ruido exógeno no lineal a la salida para verificar la coherencia
y_con_ruido = y_ideal + 0.3 * np.random.randn(len(y_ideal))

# Ejecutar análisis de coherencia
f_eje, coh = analizar_coherencia_sistema(x, y_con_ruido, fs=fs, N_fft=256, noverlap=128);
```
